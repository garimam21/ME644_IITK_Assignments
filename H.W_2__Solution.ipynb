1. Importing Libraries

```python
import numpy as np
import pandas as pd
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

2. Given data processing

```python
# Given data
temperatures = [30, 30, 30, 30, 40, 40, 40, 50, 50, 50, 60, 60, 60, 60]
hardness = [55.8, 59.1, 54.8, 54.6, 43.1, 42.2, 45.2, 31.6, 30.9, 30.8, 17.5, 20.5, 17.2, 16.9]
```

```python
# Converting lists to numpy arrays for easier manipulation
temperatures= np.array(temperatures)
hardness = np.array(hardness)
```

```python
temperatures = temperatures.reshape((-1, 1))
```

```python
e=np.empty([15])
```

3. Linear Regression Model

```python
# Creating and fitting the linear regression model
model = LinearRegression()
model.fit(temperatures, hardness)
```

4. Predicting using linear regression model

```python
# Generating predictions
# Predicted hardness values using the linear regression model
predicted_hardness = model.predict(temperatures)
w0=model.intercept_
w1=model.coef_
e=hardness-w0-w1*temperatures
print(np.sum(e))
```

5. R-squared

```python
# Calculate the coefficient of determination (R-squared)
r_squared = r2_score(hardness, predicted_hardness)
```

6. Plotting LinearRegression Line/Linear Hypothesis

```python
# Plot the original data and the linear regression line
plt.scatter(temperatures, hardness,  color='blue', label='Original Data')
plt.plot(temperatures, predicted_hardness, color='red', label='Linear Regression')
plt.xlabel('Test Temperature ($^\circ$C)')
plt.ylabel('Hardness')
plt.title('Linear Regression for Hardness vs. Temperature')
plt.legend()
plt.grid(True)
plt.show()
```

7. Coefficient of Determination

```python
print(f"Slope: {model.coef_[0]:.4f}")
print(f"Intercept: {model.intercept_:.4f}")
print(f"Coefficient of Determination (R-squared): {r_squared:.8f}")
```
